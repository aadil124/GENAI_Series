```python
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
from pydantic import BaseModel , Field

load_dotenv()

llm =ChatOpenAI(model="gpt-5")

class structured_response(BaseModel):
      name:str=Field(description="Full Name")
      email:str=Field(description="Email")
      age:int=Field(description="Age")
text= "Hi, My name is Mohd Adil Ansari, my email is ansariaadil2623@gmail.com and my Age is 32, "
structured_output = llm.with_structured_output(structured_response)
res= structured_output.invoke(f"please give me only name , eamil and age. : {text}")
res.model_dump()
print(res.name)
```
