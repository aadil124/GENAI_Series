```python
from langchain_groq import ChatGroq
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv

load_dotenv()
llm = ChatGroq(model="openai/gpt-oss-20b")
llm_openai=ChatOpenAI(model="gpt-5")

res= llm.invoke(f"what the Gen AI and Agentic AI , As I am a React Developer. If I learn Gen AI , How can we approach for a job")
print(res.content)

res1= llm_openai.invoke(f"what the Gen AI and Agentic AI , As I am a React Developer. If I learn Gen AI , How can we approach for a job")
print(res1.content)
```
